```python
import crepe.core as core

model = core.build_and_load_model(model_capacity='small')
```

```python
model.summary()
```

```python
import tensorflow as tf
import numpy as np

def representative_dataset():
    for _ in range(500):
        sample = np.random.uniform(
            -1.0, 1.0, size=(1, 1024)
        ).astype(np.float32)
        yield [sample]

fixed_input = tf.keras.Input(shape=model.input_shape[1:], batch_size=1)
fixed_model = tf.keras.Model(inputs=fixed_input, outputs=model(fixed_input))

converter = tf.lite.TFLiteConverter.from_keras_model(fixed_model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_dataset
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.int8
converter.inference_output_type = tf.int8

tflite_model = converter.convert()
```

```python
with open('crepe_small_int8.tflite', 'wb') as f:
    f.write(tflite_model)
```

```python
import tensorflow as tf
interpreter = tf.lite.Interpreter(model_path='crepe_small_int8.tflite')
interpreter.allocate_tensors()

input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print(output_details)
```
